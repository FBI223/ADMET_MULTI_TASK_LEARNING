```python
import pandas as pd

# Ścieżka do Twojego pliku
file_path = 'dataset.parquet'

# Wczytanie danych
try:
    # Pandas automatycznie rozpozna silnik, jeśli masz pyarrow/fastparquet
    df = pd.read_parquet(file_path)
    
    print("=== PODSTAWOWE INFO ===")
    print(df.info()) # Typy danych, liczba niepustych rekordów (non-null)

    print("=== DATASETY ===")
    print(df['task'].unique())

    print("\n=== LISTA KOLUMN (WŁAŚCIWOŚCI) ===")
    print(df.columns.tolist())
    
    print("\n=== STATYSTYKI OPISOWE ===")
    # Dobre do sprawdzenia zakresów wartości (min, max, mean) dla regresji ADMET
    print(df.describe())
    
    print("\n=== PIERWSZE 5 WIERSZY ===")
    print(df.head())

    # Jeśli chcesz sprawdzić ile masz brakujących danych (częste w TDC)
    print("\n=== BRAKUJĄCE DANE (NaN) ===")
    print(df.isnull().sum())

except Exception as e:
    print(f"Błąd podczas wczytywania: {e}")
    print("Upewnij się, że masz zainstalowane: pip install pyarrow")
```
